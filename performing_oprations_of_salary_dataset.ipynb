```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv("salary.data.csv")
print(df)
```

```python
df.head()
```

```python
X = df.iloc[:,0:1].values
y = df.iloc[:,1].values

print("X values")
print(X)

print("Y values")
print(y)
```

```python
lr = LinearRegression()

lr.fit(X,y)
y_pred = lr.predict(X)
print("R2 Score =", r2_score(y, y_pred))
```

```python
poly2 = PolynomialFeatures(degree=2)

X_poly2 = poly2.fit_transform(X)

model2 = LinearRegression()

model2.fit(X_poly2,y)

pred2 = model2.predict(X_poly2)

print("Degree 2 R2 Score =",r2_score(y,pred2))
```

```python
poly3 = PolynomialFeatures(degree=3)

X_poly3 = poly3.fit_transform(X)

model3 = LinearRegression()

model3.fit(X_poly3,y)

pred3 = model3.predict(X_poly3)

print("Degree 3 R2 Score =",r2_score(y,pred3))
```

```python
poly4 = PolynomialFeatures(degree=4)

X_poly4 = poly4.fit_transform(X)

model4 = LinearRegression()

model4.fit(X_poly4,y)

pred4 = model4.predict(X_poly4)

print("Degree 4 R2 Score =",r2_score(y,pred4))
```

```python
print("Linear Regression :",r2_score(y,y_pred))

print("Degree 2 :",r2_score(y,pred2))

print("Degree 3 :",r2_score(y,pred3))

print("Degree 4 :",r2_score(y,pred4))
```

```python
plt.scatter(X,y)

plt.plot(X,y_pred)

plt.title("Linear Regression")

plt.xlabel("Level")

plt.ylabel("Salary")

plt.show()
```

```python
X_grid=np.arange(min(X),max(X),0.1)

X_grid=X_grid.reshape(len(X_grid),1)

plt.scatter(X,y)

plt.plot(X_grid,
         model4.predict(poly4.fit_transform(X_grid)))

plt.title("Polynomial Regression Degree 4")

plt.xlabel("Level")

plt.ylabel("Salary")

plt.show()
```
